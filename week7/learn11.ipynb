# 내장함수

학습목표
- 파이썬에 내장된 내장 함수들을 활용할 수 있다.
- 람다식을 활용할 수 있다.
- 이터레이터와 제너레이터를 시용하여 반복 가능한 객체를 작성할 수 있다.
- 연산자 중복(Operation Overloading)을 하는 클래스를 작성할 수 있다.
- 모듈의 개념을 이해하고 적용할 수 있다.
- 유용한 모듈을 설치하고 사용할 수 있다.

abs() : 절대값 반환

all() : 시퀀스(리스트, 튜플, 집합, 사전)를 입력받아 모든 항목이 참이면 True를 반환함

any() : 시퀀스 목록을 입력 받아 하나라도 참이면 True를 반환함

eval() : 전달된 문자열을 파이썬 수식문장으로 생각하여 해독하고 수식을 실행한다. 동적 프로그래밍 가능케 하는 함수임

sum() : 리스트 원소가 숫자(int, float)일 때 항목들을 더하여 합을 반환

len() : 객체 원소의 길이(갯수)를 반환한다.

list(), set(), dict(), tuple() : 각 타입의 비어있는 시퀀스 객체를 생성하여 반환한다.

map(함수, 반복객체) : 반복객체의 각 원소에 대해서 함수를 수행한 결과 맵 시퀀스 객체를 반환한다.

dir(객체 또는 클래스) : 객체/클래스가 가지고 있는 변수나 함수를 출력함

max(), min() : 리스트, 튜플, 문자열 등에서 가장 큰 원소/가장 작은 원소를 반환한다.

enumerate () : 시퀀스 객체를 입력받아 열거형 객체를 반환한다. (인덱스, 객체) 원소 형태의 enumerate객체를 반환

filter(필터함수, 시퀀스) : 원소 중 필터함수를 만족하는 원소만 추린 filter객체를 반환한다. 보통 list로 형변환하여 사용

zip(리스트1, 리스트2) : 각 리스트의 동일한 인덱스를 갖는 원소끼리 튜플로 묶은 zip객체를 반환. 길이가 다를 때는 짧은 길이만큼만 수행.

sorted(시퀀스, key함수, reverse여부) : 시퀀스 객체를 받아 원소를 정렬하여 정렬된 시퀀스 객체를 생성하고 반환한다.

- (key함수는 각 원소를 입력받아 비교할 크기를 반환하는 함수, reverse=True이면 내림차순)

int(), str(), float() : 형변환 함수

```python
class Person(object):
    def __init__(self, name, age):
             self.name = name
             self.age = age
    def __repr__(self):
             return "<이름: %s, 나이: %s>" % (self.name, self.age)
    def getName(self):
             return self.name

def keyAge(person):  # 크기를 비교할 값 (나이)
     return person.age

people = [Person("윤종하", 20), Person("한찬승", 35), Person("윤완", 38)]
print(sorted(people, key = keyAge))
print(sorted(people, key = lambda x: x.getName()))
```

# lambda 함수 : 무명함수
- 함수 내부문장이 간단하여 함수로 정의하지 않고 간단히 쓰고 싶을 때 사용
    - 문법: lambda x : x ** 2 --> 입력 x를 받아 제곱한 후 반환
    - 예: list1 = list(range(5), map(lambda x: x**2, list1) --> 각 항목을 제곱한 map객체를 반환
- 이름을 붙일 수도 있음
    - func1 = lambda x : x ** 2 --> def func1(x): return x**2 과 같음

# 이터레이터(iterator)
- for 루프와 함께 사용할 수 있는 객체들을 iterable 객체라고 한다.

- iterable객체는 iter() 함수로 iterator 객체를 반환하고 iterator.next() 함수를 제공해야 한다.

- 제공해야 하는 메소드

    - iter() : iterator 객체를 반환
    - next() : 다음 원소 반환, 더 이상 원소가 없으면 StopIteration 예외를 발생하도록 한다.
- 시퀀스 객체(list, dict, set, tuple) 은 이터러블 객체이다.

- enumerate, zip, range 등도 이터러블 객체이다.

- 사용자 클래스가 이터러블 하려면 위 두 메소드를 정의해야 한다.

```python
class MyCounter(object):
    # 생성자 메소드를 정의한다. 
    def __init__(self, low, high):
        self.current = low
        self.high = high

    # 이터레이터 객체로서 자신을 반환한다. 
    def __iter__(self):
        return self

    def __len__(self):
        return self.high - self.current + 1

    def __next__(self):
        # current가 high 보다 크면 StopIteration 예외를 발생한다.
        # current가 high 보다 작으면 다음 값을 반환한다. 
        if self.current > self.high:
            raise StopIteration
        else:
            self.current += 1
            return self.current - 1

mc = MyCounter(1, 10)
for x in mc:
    print(x)
```

### Generator
- yield 키워드를 사용하여 임의의 반복가능한 원소들을 클래스에 임의로 정의한다.
- 자동으로 next()와 iter 가 정의된다.

```python
def myGenerator():
     yield 'first'
     yield 'second'
     yield 'third'

for word in myGenerator():
     print(word)
```
