- 캡슐화 : 데이터와 연산을 하나의 덩어이로 묶어서 다룸
- 다형성 : 함수 이름이 같더라도 문맥(혹은 클래스)에 따라 다른 함수가 수행됨

상속의 예
- 큰 주제(부모클래스) 밑으로 연관되어있는 작은 주제들로 이루어져있다.(자식클래스)
- Animal(동물)
    - Lion(사자), Dog(개), Cat(고양이)
- Vehicle(탈것)
    - Car(자동차), Bus(버스), Boat(보트)

```python
# 부모클래스 Car
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker                      # 메이커
        self.model = model                      # 모델
        self.color = color                      # 색상
        self.price = price                      # 가격


    def setMaker(self, maker):
        self.maker = maker


    def getMaker(self):
        return self.maker


    def getDesc(self):
       return self.__str__()


    def __str__(self):
        return f"Maker : {self.maker}, Model : {self.model}, Color : {self.color}, Price : {self.price}"

# 자식클래스 ElectricCar
class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterySize = bat_size

    def getbatterySize(self):
        return self.batterySize

    def setBatterySize(self, bat_size):
        self.batterySize = bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery : {self.batterySize}kh"

    def getDesc(self):
        return self.__str__()

# 자식클래스 Truck
class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad):
        super().__init__(maker, model, color, price)
        self.payLoad = payLoad

    def __str__(self):
        return f"{super().__str__()}, PayLoad : {self.payLoad}kg"

# 테스트 car1
car1 = Car("kia", "Ray", "Blue", 2000)
print(car1.getDesc())
print(car1.model)
# 테스트 ecar1
ecar1 = ElectricCar("Tesla", "Model S", "Red", 8000, 81)
print(ecar1)
ecar1.setBatterySize(92)
print(ecar1.getDesc())
# 테스트 truck1
truck1 = Truck("Ford", "F-158", "Black", 4000, 1000)
print(truck1)
print(truck1.getMaker())        # getMaker가 없지만 자식클래스가 부모클래스한테 상속받아 사용가능 또 private 속성도 상속받을 수 있지만 볼수도 없고 수정할수도 없다. 

# 테스트 type 및 상속
print(type(car1))
print(type(ecar1))
print(type(truck1))

print(isinstance(car1, Car))
print(isinstance(ecar1, ElectricCar))
print(isinstance(truck1, Truck))

print(isinstance(ecar1, Car))
print(isinstance(truck1, Car))

#print(isinstance(truck1, Vehicle))
```

# 가시성 : public, protected(_로 시작), vate(__로 시작)


```python
class Parent(object):
    def __init__(self):
        self.__private_money = 1000
        self._protected_money = 500
        self.public_money = 100
    
    def __str__(self):
        return f"Parent has total {self.__private_money + self._protected_money + self.public_money} money"
    
class Child(Parent):
    def __init__(self):
        super().__init__()
    
    def __str__(self):
        return f"Child : {self._protected_money + self.public_money} public money"
    
def test():
    p = Parent()
    print(p)
    c = Child()
    print(c)
    
    print(p.public_money)
    print(p._protected_money)
#    print(p.__private_money)
    
    print(f"Child : {c.public_money}")
    print(f"Child : {c._protected_money}")
#    print(f"Child : {c.__private_money}")
  
test()
```

# 다중 상속: 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self, v1 = 10):
        self.attr1 = v1
    
    def display1(self):
        print(f"Base1: {self.attr1}")

    def display(self):
        self.display1()
        
class Base2:
    def __init__(self, v2 = 20):
        self.attr2 = v2
    def display2(self):
        print(f"Base2: {self.attr2}")
        
    def display(self):
        self.display2()
        
class C(Base1, Base2):
    def __init__(self, v1, v2, v3):
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.attr3 = v3
        
    def print(self):
        print(f"{self.attr1}, {self.attr2}, {self.attr3}")
        self.display1()
        self.display2()
        self.display()
        Base1.display(self)
        Base2.display(self)
        
def test():
    c = C(v1=1, v2=2, v3=3)
    c.print()
    
    c2 = C(4,5,6)
    c2.print()
    
if __name__ == "__main__":
    test()
```

Overriding : 부모의 함수(메소드) 재정의

```python
import math

class Shape:
    def __init__(self, name="도형"):
        self.name = name
        
    def draw(self):
        print(f"Shape:{self.name}을(를) 그립니다.")
        
    def get_area(self):
        print(f"Shape:{self.name}의 면적을 계산합니다")

class Circle(Shape):
    def __init__(self, radius=1, name="원"):
        super().__init__(name)
        self.radius = radius
    def draw(self):
        print(f"Circle:{self.name}을(를) 그립니다.")
        
    def get_area(self) -> float:
        print(f"Circle:{self.name}의 면적을 계산합니다")
        return math.pi * self.radius ** 2
    
class Rectangle(Shape):
    def __init__(self, width = 1, height = 1, name = "사각형"):
        super().__init__(name)
        self.width = width
        self.height = height

    def draw(self):
        print(f"Rectangle:{self.name}을(를) 그립니다.")
        
    def get_area(self) -> float:
        print(f"Rectangle:{self.name}의 면적을 계산합니다")
        return self.width * self.height
    
def test():
    c = Circle(10)
    c.draw()
    print(c.get_area())
    
    r1= Rectangle(width = 5, height=10)
    r1.draw()
    print(r1.get_area())
    
    print("-------------------------------")
    circles = [Circle(1), Circle(2), Circle(3)]
    for c in circles:
        #c.draw()
        print(c.get_area())
    print("-------------------------------")

    print("-------------------------------")
    rectangles = [Rectangle(1,1), Rectangle(2,2), Rectangle(3,3)]
    for r in rectangles:
        #r.draw()
        print(r.get_area())
    print("-------------------------------")

    print("-------------------------------")
    shapes = [Circle(1), Rectangle(1,1), Circle(20), Rectangle(2,3)]
    for s in shapes:
        #s.draw()
        print(s.get_area())
    print("-------------------------------")
    
if __name__ == "__main__":
    test()
```
