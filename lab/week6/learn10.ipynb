## 파일 읽고 쓰기

| 파일 모드 | 모드 이름 | 설명 |
| --- | --- | --- |
| "r" | 읽기 모드 | 파일을 처음부터 읽는다 |
| "w" | 쓰기 모드 | 파일의 처음부터 씀, 없으면 생성,만약 존재하면 기존내용 지워지고 새로 작성 |
| "a" | 추가 모드 | 파일의 끝에 쓴다. 없으면 생성 |
| "r+" | 읽기 쓰기 모드 | 읽기와 쓰기 모드 |

# 작업폴더 : 파이썬 파일을 실행하는 위치
- 가변길이 레코드(한 라인,한 줄) : 예시 이름 여러자 입력가능 3자리 2자리 가 존재함

```python
with open("file.txt", "r", encoding="utf-8") as f:
   # for line in f:
   #     print(line.strip())     
    line = f.readline().strip()
    while line != "":
        print(line)
        line = f.readline().strip()
        
    eline = f.readline().strip()
    print(eline)
    print(f.tell())
    
    f.seek(0,0)
    eline = f.readline().strip()
    print(eline)
    print(f.tell())
```
